# 1. Exploratory Data Analysis

This notebook loads the Criteo Uplift Dataset and summarizes:
- Dataset shape and key statistics
- Treatment/control group distribution  
- Outcome variable distributions (conversion, visits)
- Feature summary statistics

The goal: Understand the data before statistical testing.

```python
import pandas as pd
import numpy as np

import os
DATA_PATH = os.path.join("data", "raw", "criteo-uplift-v2.1.csv.gz")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
print("Dataset shape:", df.shape)

print("\nTreatment distribution:")
print(df["treatment"].value_counts())

print("\nOutcome totals:")
print("Conversions:", df["conversion"].sum())
print("Visits:", df["visit"].sum())
print("Exposures:", df["exposure"].sum())
```

```python
group_metrics = (
    df.groupby("treatment")
      .agg(
          observations=("treatment", "size"),
          conversions=("conversion", "sum"),
          visits=("visit", "sum"),
          exposures=("exposure", "sum")
      )
      .reset_index()
)

group_metrics["conversion_rate"] = (
    group_metrics["conversions"] / group_metrics["observations"]
)

group_metrics["visit_rate"] = (
    group_metrics["visits"] / group_metrics["observations"]
)

group_metrics
```

```python
feature_summary = []

for col in [f"f{i}" for i in range(12)]:
    feature_summary.append({
        "feature": col,
        "unique_values": df[col].nunique(),
        "missing_values": df[col].isna().sum(),
        "most_common_share": df[col].value_counts(normalize=True).iloc[0]
    })

feature_summary = pd.DataFrame(feature_summary)

feature_summary
```
